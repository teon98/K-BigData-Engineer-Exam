[제공 데이터]
- 공부시간, 수면시간, 합격여부(0:불합격, 1:합격)

[문제]
1. 공부시간, 수면시간을 독립변수로 하여 학습용(train) 데이터를 이용해 로지스틱 회귀분석을 수행하고, 공부시간이 3 증가할 때 오즈비가 얼마나 증가하는지 구하여라.

2. ①에서 구축한 모형의 잔차이탈도(Deviance)를 구하여라.

3. ①에서 구축한 모형을 시험용(test) 데이터에 적용하여 예측값을 구하고, 정확도(accuracy)를 구하여라.

# 풀이

```python
import statsmodels.formula.api as smf
import numpy as np

# 1) 학습/평가 분할
train, test = df.iloc[:40], df.iloc[40:]

# 2) 로지스틱 회귀 적합(규제 없음)
model = smf.logit("합격여부 ~ 공부시간 + 수면시간", data=train).fit()

# (1) 공부시간이 3 증가할 때 오즈비 배율
print(np.exp(model.params))
odds_ratio_3 = float(np.exp(model.params["공부시간"] * 3))
print("오즈비(공부시간 3 증가):", round(odds_ratio_3, 3))

# (2) 잔차이탈도 = -2 * 로그우도
deviance = -2 * float(model.llf)
print("잔차이탈도:", round(deviance, 3))

# (3) test 정확도
from sklearn.metrics import accuracy_score
proba = model.predict(test)
pred  = (proba >= 0.5).astype(int)
acc   = accuracy_score(test['합격여부'], pred)
print("정확도:", round(acc, 3))
```
