## 결측값 개념과 탐색

### 결측값이란?
- 데이터셋에서 특정값이 누락된 상태
- 결측값은 분석, 시각화, 모델링 등에 오류 또는 왜곡을 일으킬 수 있음

```python
import pandas as pd

data = {
    '이름': ['철수', '영희', '민수', '선호'],
    '나이': [25, None, 30, 28],
    '성별': ['남', '여', '남', None]
}

df = pd.DataFrame(data)
df
```

#### 결측값 탐색
1. `isnull()`
- 각 원소가 결측값인지 아닌지 Boolean(True/False)로 반환

```python
df.isnull()
```

2. `isnull().sum()`
- 열 단위로 결측값 개수 집계
- (참고) True: 1 , False: 0

```python
df.isnull().sum()
```

### 결측값 채우기
#### (1) `fiilna()` 함수를 활용하여, 결측값을 원하는 값으로 채우기

#### 1. 특정 값으로 채우기

```python
df['나이'].fillna(0)
```

#### 2. 열의 평균값으로 채우기

```python
df['나이'].fillna(df['나이'].mean())
```

#### 3. 열의 최빈값으로 채우기

```python
df['성별'].fillna(df['성별'].mode()[0])
```

#### (참고) mode() 함수에 대하여
- mode에는 '유행' 이라는 의미가 있음.
- 데이터 중에서 가장 유행하는 값을 가져와라 라는 뜻

왜 mode() 함수를 쓰고 [0]을 가져올까?
- mode() 함수를 쓰면 Series(`{index: value}`)의 형태로 뱉어냄 
- (참고) 여러 개의 최빈 값이 존재할 수 있기 때문에 Series 형태로 반환 함.
- 그래서 return 된 최빈값 중 가장 첫번째 값으로 채우자! 라는 의미로 [0]을 붙여줌

```python
df['성별'].mode()
```

```python
df['성별'].mode()[0]
```

#### (2) `bfill`, `ffill` 함수로 이전/다음 값으로 채우기

#### 1. 이전 값으로 채우기 : `ffill()`
(의미) Forward Fill

```python
df
```

```python
df.ffill()
```

#### 2. 다음 값으로 채우기: `bfill()`
(의미) Backward Fill

```python
df
```

```python
df.bfill()
```

### 결측값 제거

#### `dropna()` 함수로 결측값 제거
(참고) na의 의미: Not Available(NA) 이용할 수 없는
#### 1. 결측값 포함된 행 전체 제거

```python
df
```

```python
df.dropna()
```

#### 2. 특정 열 기준으로 제거

```python
df.dropna(subset=['성별']) #subset: 부분집합
```
