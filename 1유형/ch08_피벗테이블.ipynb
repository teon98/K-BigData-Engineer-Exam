```python
import pandas as pd

data = {
    'Department': ['Sales', 'Sales', 'HR', 'HR', 'Sales', 'IT', 'IT'],
    'Gender': ['M', 'F', 'M', 'F', 'M', 'M', 'F'],
    'Salary': [3000, 3200, 2500, 2700, 3100, 4000, 3900]
}

df = pd.DataFrame(data)
```

## 피벗테이블(Pivot Table) 이란?
- 데이터를 요약(집계)하고 차원 축소하여 가독성 높은 형태로 재구성하는 도구
- 행 인덱스(index), 값(values)을 지정하여 집계 함수(aggfunc)로 요약
> groupby -> 단순 집계에 강함. (시험 관점에서는 피벗테이블을 몰라도 된다)

<br>

`pd.pivot_table(index='컬럼A', columns='컬럼B', values='컬럼C', aggfunc='mean')`
| 매개변수 | 설명 |
| ------ | --- |
| Index | 행 인덱스로 사용할 컬럼 |
| columns | 열 이름으로 사용할 컬럼 | 
| values | 집계 대상 컬럼 | 
| aggfunc | 집계 함수 (기본값: mean)|

### 1. 단순 피벗 테이블

```python
pd.pivot_table(df, index='Department', values='Salary', aggfunc='mean')
```

### 2. 다중 피벗 테이블

```python
pd.pivot_table(df, index='Department', columns='Gender', values='Salary', aggfunc='mean')
```

### 3. 여러 집계 함수 사용

```python
pd.pivot_table(df, index='Department', columns='Gender', values='Salary', aggfunc=['mean', 'max', 'min'])
```

> ### 스택(Stack)과 언스택(UnStack)
> - stack(): wide -> long 형태
> - unstack(): long -> wide 형태

```python
pivot = pd.pivot_table(df, index='Department', columns='Gender', 
                       values='Salary', aggfunc=['mean', 'max', 'min'])
pivot
```

```python
pivot1 = pivot.stack()
pivot1
```

```python
pivot2 = pivot1.unstack()
pivot2
```
