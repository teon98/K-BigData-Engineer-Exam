### 하이퍼파라미터 튜닝
#### 1. 시험에서 튜닝할만한 하이퍼파라미터
| 하이퍼파라미터 | 의미 | 유의사항 |
| ---------- | --- | ------ |
| n_estimators | 트리 개수, 성능에 영향이 큼 | 너무 올리면 시간 초과 가능 |
| max_depth | 트리 최대 깊이, 과적합 조절 | 깊을수록 과적합 가능성 |

#### 2. 추천하는 하이퍼파라미터 값
- n_estimators = 100, 300, 500
- max_depth = 4, 6, 8

```python
RandomForestClassifier(n_estimators=300, max_depth=6)
XGBClassifier(n_estimators=300, max_depth=6)
LGBMClassifier(n_estimators=300, max_depth=6)
```

#### 3. 기타 활용가능 파라미터 (Boosting 계열)
| 하이퍼파라미터 | 의미 | 추천 값 |
| ---------- | --- | ------ |
| learning_rate | 학습률 | 0.01 ~ 0.2 |
| subsample | 데이터 샘플링 비율 | 0.6 ~ 1.0 |
