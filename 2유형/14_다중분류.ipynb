## 다중 분류 데이터셋 분할
### 평가지표
| 지표 | 설명 | 해석 |
| --- | --- | --- |
| ROC_AUC | ROC 커브의 아래 면적 <br> (양성/음성 분류의 종합 성능 지표) | 1에 가까울수록 좋음(0.5=랜덤) |
| Accuracy | 전체 예측 중 정답 비율 | 높을 수록 좋음 |
| Precision | 양성 예측 중 실제 양성 비율 | 높을 수록 좋음 |
| Recall | 실제 양성 중 모델이 맞춘 비율 | 높을 수록 좋음 |
| F1 Score | 정밀도와 재현율의 조화 평균 | 1에 가까울 수록 좋음 |

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

# 예측값
y_pred = model.predict(X_val)
y_proba = model.predict_proba(X_val)

# 평가지표
acc = accuracy_score(y_true, y_pred)
prec = precision_score(y_true, y_pred, average='macro') # 또는 'weighted'
rec = recall_score(y_true, y_pred, average='macro')
f1 = f1_score(y_true, y_pred, average='macro')
auc = roc_auc_score(y_true, y_pred, multi_class='ovr', average='macro')
```

- average='macro': 클래스별 성능 동일 가중 평균 (클래스 더 많은 애들한테 더 많은 가중을 줄꺼니)
- average='weighted': 샘플 수 기반 평균
- multi_class='ovr: One-vs-Rest 방식의 ROC AUC (그룹 A,B,C,D 있으면 A vs B,C,D 할꺼니)
