### DataFrame 생성
#### (1) 2차원 표에서 **컬럼 단위**로 데이터를 표현
- 컬럼명을 딕셔너리의 key로 데이터는 딕셔너리의 values로 사용

```python
import pandas as pd

data = {
    "종가": [157000, 51300, 68800, 140000],
    "PER": [39.88, 8.52, 10.03, 228.38],
    "PBR": [4.38, 1.45, 0.87, 2.16]
}

index = ['NAVER', '삼성전자', 'LG전자', '카카오']

df = pd.DataFrame(data, index)
print(df)
```

#### (2) 2차원 표에서 **로우 단위**로 데이터를 리스트로 표현
- data, index, columns를 각각 리스트로 표현

```python
import pandas as pd

data = [
    [157000, 39.88, 4.38],
    [51300, 8.52, 1.45],
    [68800, 10.03, 0.87],
    [140000, 228.38, 2.16]
]

index = ['NAVER', '삼성전자', 'LG전자', '카카오']
columns = ['종가', 'PER', 'PBR']
df = pd.DataFrame(data=data, index=index, columns=columns)
print(df)
```

- data에 column 명을 kye:value를 가진 딕셔너리로 표현하여 생성하는 방법

```python
import pandas as pd

data = [
    {"종가": 15700, "PER": 39.88, "PBR": 4.38},
    {"종가": 51300, "PER": 8.52, "PBR": 1.45},
    {"종가": 68800, "PER": 10.03, "PBR": 0.87},
    {"종가": 140000, "PER": 228.38, "PBR": 2.16},
]

index = ['NAVER', '삼성전자', 'LG전자', '카카오']
df = pd.DataFrame(data=data, index=index)
print(df)
```

### 연습문제
#### Q. 다음 2차원 데이터를 데이터프레임으로 생성하세요.
| 암호화폐 | 시가 | 고가 | 저가 | 종가 |
| ------ | --- | --- | --- | --- | 
| 비트코인 | 980 | 990 | 920 | 930 | 
| 리플 | 200 | 300 | 180 | 180 |
| 이더리움 | 300 | 500 | 300 | 400 |

```python
# 첫번째 방법: 컬럼명을 딕셔너리의 키로 데이터를 values로 dataframe을 표현하는 방법
import pandas as pd

data = {
    "시가" : [980, 200, 300],
    "고가" : [990, 300, 500],
    "저가" : [920, 180, 300],
    "종가" : [930, 180, 400]
}

index = ["비트코인", "리플", "이더리움"]
df = pd.DataFrame(data = data, index = index)
print(df)
```

```python
# 두 번째 방법: 인덱스와 컬럼명을 따로 지정하는 방법
import pandas as pd

data = [
    [980, 990, 920, 930],
    [200, 300, 180, 180],
    [300, 500, 300, 400]
]

index = ['비트코인', '리플', '이더리움']
columns = ['시가', '고가', '저가', '종가']

df = pd.DataFrame(data = data, index=index, columns=columns)
print(df)
```

```python
# 세 번째 방법: 컬럼명:value 를 가진 딕셔너리로 data를 구성하는 방법
import pandas as pd

data = [
    {"시가": 980, "고가": 990, "저가": 920, "종가": 930},
    {"시가": 200, "고가": 300, "저가": 180, "종가": 180},
    {"시가": 300, "고가": 500, "저가": 300, "종가": 400},
]

index = ['비트코인', '리플', '이더리움']

df = pd.DataFrame(data, index)
print(df)
```
