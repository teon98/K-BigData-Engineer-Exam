### DataFrame 생성
#### (1) 2차원 표에서 **컬럼 단위**로 데이터를 표현
- 컬럼명을 딕셔너리의 key로 데이터는 딕셔너리의 values로 사용

```python
import pandas as pd

data = {
    "종가": [157000, 51300, 68800, 140000],
    "PER": [39.88, 8.52, 10.03, 228.38],
    "PBR": [4.38, 1.45, 0.87, 2.16]
}

index = ['NAVER', '삼성전자', 'LG전자', '카카오']

df = pd.DataFrame(data, index)
print(df)
```

#### (2) 2차원 표에서 **로우 단위**로 데이터를 리스트로 표현
- data, index, columns를 각각 리스트로 표현

```python
import pandas as pd

data = [
    [157000, 39.88, 4.38],
    [51300, 8.52, 1.45],
    [68800, 10.03, 0.87],
    [140000, 228.38, 2.16]
]

index = ['NAVER', '삼성전자', 'LG전자', '카카오']
columns = ['종가', 'PER', 'PBR']
df = pd.DataFrame(data=data, index=index, columns=columns)
print(df)
```
