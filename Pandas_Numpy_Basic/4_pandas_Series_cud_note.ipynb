### Series에 값 추가하기
딕셔너리와 유사한 방식으로 값 추가 가능
- `s.loc[인덱스] = 값`

```python
import pandas as pd

data = [100, 200, 300]
index = ["월", "화", "수"]
s = pd.Series(data, index)
print(s)
```

```python
s["목"] = 400
print(s)
```

### Series 삭제
drop 메서드 사용
- 원본은 유지하고 값이 삭제된 시리즈 객체를 리턴
- `s.drop('인덱스')`
- `s.drop(['인덱스1', '인덱스2'])`

```python
data2 = [100, 200, 300]
index2 = ["월", "화", "수"]
s2 = pd.Series(data2, index2)
print(s2)
```

```python
s2_1 = s2.drop("월")
print(s2_1)
print(s2) #원본 유지
```

(참고) 원본 값을 삭제하고 싶을 때 → `inplace=True` 설정

```python
print(s2)
s2.drop("수", inplace=True)
print(s2)
```

(참고) 여러 값을 동시에 삭제하고 싶을 때 

```python
s2.drop(["월","화"])
```
