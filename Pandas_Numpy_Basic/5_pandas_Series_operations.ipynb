## 브로드캐스팅(Broadcasting)
반복문을 사용하지 않아도 연산이 시리즈 객체의 전체 값에 적용됨

> 브로드캐스팅 뜻
> 모양이 다른 두 배열(Series/DataFrame)간에 산술 연상이 가능하도록 크기를 자동으로 맞춰주는 기능
> 원래는 행과 열의 개수가 딱 맞아야 원소별 연산이 가능하지만, 부족한 쪽 데이터의 형태를 알아서 '늘려서' 매칭해주는 기능.
> 
> <img src="https://www.boardinfinity.com/blog/content/images/2023/02/Numpy-Broadcasting.png" width="400px">

````python
import pandas as pd

s = pd.Series([100, 200, 300])
print(s+10)
````

## 사칙 연산
반복문을 사용하지 않아도 인덱스를 기준으로 값끼리 사칙 연산이 적용됨.

````python
# 인덱스가 지정되어 있지 않은 경우
high = pd.Series([51500, 51200, 52500, 51500, 51500])
low = pd.Series([50700, 50500, 50500, 50800, 50700])
diff = high-low
print(diff)
````

````python
# 명시적인 인덱스가 지정되어 있는 경우
high2 = pd.Series([51500, 51200, 52500], index=["5/1", "5/2", "5/3"])
low2 = pd.Series([50700, 50500, 50500], index=["5/1", "5/2", "5/4"])
diff2 = high2-low2
print(diff2)
````

## 시리즈 비교 연산
비교 연산자의 결과로 Boolean 타입이 저장된 시리즈가 리턴됨

````python
s = pd.Series(data=[100, 200, 300, 400, 500])
cond = s > 300
print(cond)
````

## 시리즈 필터링
True/False 값을 통해서 True 값만 필터링 가능

````python
s = pd.Series(data=[100, 200, 300, 400, 500])
cond = [False, False, False, True, True]
s[cond]
````

````python
cond2 = s>400
print(s[cond2])
````

````python
# 시가가 300원 이상인 날의 종가를 출력
시가 = pd.Series([100, 200, 300, 400, 500])
종가 = pd.Series([100, 300, 400, 500, 600])
조건 = (시가 >= 300)
print(종가[조건])
````

## 연습문제
#### Q1. 고가와 저가 차이가 100 이상인 날의 고가를 출력하세요.
```python
저가 = Series([10, 200, 200, 400, 600])
고가 = Series([100, 300, 400, 500, 600])
```

````python
import pandas as pd

저가 = pd.Series([10, 200, 200, 400, 600])
고가 = pd.Series([100, 300, 400, 500, 600])

조건 = 고가 - 저가 >= 100
print(고가[조건])
````

#### Q2. 종가가 80000원 이상 90000원 미만인 날짜를 출력하세요.
```python
종가 = Series([93000, 82400, 99100, 81000, 72300], ['05/14', '05/15', '05/16', '05/17', '05/18'])
```

````python
import pandas as pd
종가 = pd.Series([93000, 82400, 99100, 81000, 72300], ['05/14', '05/15', '05/16', '05/17', '05/18'])
조건 = (종가 >= 80000) & (종가 < 90000) #파이썬에서 and, or, not은 pandas에서 &, |, ~ 을 써야한다
print(종가[조건].index)
````
