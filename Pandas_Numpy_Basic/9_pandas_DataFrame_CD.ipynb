### 컬럼 삭제
- `df.drop('컬럼명', axis=1)`

> inplace의 의미는 '제자리에', '본래 있던 곳에' 라는 의미로, <br>
> 제자리에 있던 곳에 해당 명령어를 수행하라~ 라는 의미

### 시계열 데이터와 인덱스
시계열 데이터는 인덱스가 날짜와 시간으로 구성됨.
- 문자열로 표현된 날짜와 시간을 DatetimeIndex 타입으로 변환해서 사용해야 함.

```python
# 판다스의 to_datetime으로 문자열을 datetimeindex 타입으로 변환하기
import pandas as pd

date = ["2021-04-09"] # 날짜가 여러개일 때
index = pd.to_datetime(date)

print(type(date))
print(index)
print(type(index))
```

```python
date = "2021-04-09" # 날짜가 한 개 일 때
index = pd.to_datetime(date)

print(type(date))
print(index)
print(type(index))
```
