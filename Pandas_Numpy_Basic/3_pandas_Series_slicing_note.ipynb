## Series 인덱싱
### ▪︎ iloc 속성(변수)를 사용하여 인덱싱 하기
시리즈 객체의 행 번호를 사용하여 인덱싱

(참고) iloc 뜻
integer-based location 

```python
import pandas as pd

data = [100, 200, 300]
index = ["월", "화", "수"]
s = pd.Series(data, index)
print(s)
print(s.iloc[0])
print(s.iloc[2])
print(s.iloc[-1])
```

### ▪︎ loc 속성을 사용하여 인덱싱 하기
시리즈 객체의 인덱스로 인덱싱

(참고) loc 뜻 label-based location (or location)

```python
print(s.loc["월"])
print(s.loc["화"])
print(s.loc["수"])
```

### ▪︎ 대괄호([])를 사용하여 인덱싱 하기
- 기본적으로 인덱스를 기준으로 인덱싱 가능 (간단하게 인덱싱하고 싶을 때 사용)
- 하지만 권장하는 방법은 아니며 명시적으로 iloc나 loc를 사용하기

```python
print(s["월"])
print(s.loc["월"])
```

## Series 슬라이싱
### ▪︎ 행번호를 사용하여 슬라이싱 하기
- iloc[시작 행번호:끝 행번호]

```python
print(s.iloc[0:2])
```

### ▪︎ 인덱스를 사용하여 슬라이싱 하기
- loc[시작 인덱스:끝 인덱스]
- (주의) loc를 쓸 대는 행 번호 인덱싱과 다르게 끝 인덱스가 포함됨!

```python
print(s.loc["월":"화"])
```

### ▪︎ 여러 값 인덱싱 하기
연속적이지 않은 여러 개의 값을 한 번에 인덱싱하기
- 행 번호 또는 인덱스를 파이썬 리스트로 구성하고,
- loc 또는 iloc 속성에서 해당 리스트 사용

```python
target = [0,2]
print(s.iloc[target])
```

```python
target = ["월", "수"]
print(s.loc[target])
```
