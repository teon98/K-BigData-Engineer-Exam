## 파이썬 및 NumPy 자료구조

| 자료구조 | 특징 | 예제코드 |
| -----  | -- | ------ |
| 리스트 | 순서가 있고 수정 가능 | `[1,2,3]` |
| 튜플 | 순서가 있고 수정 불가능 | `[(1,2,3)`]|
| 딕셔너리 | 연관된 데이터 (key:value) 쌍을 저장 | `{'coffee': 5000, 'bread':2000}` |
| NdArray | 반복문 없이 데이터 배열에 대한 처리 및 선형 대수와 관련된 기능 제공 | `arr = np.array([1,2,3])` |

## Pandas 기본 자료구조
| 자료구조 | 특징 | 
| -- | -- |
| Series | NumPy를 기반으로 만들어진 1차원 데이터를 위한 자료구조 | 
| DataFrame | NumPy를 기반으로 만들어진 2차원 데이터를 위한 자료구조 |

## Series 객체 생성하기
1차원 데이터를 위한 자료구조
- Raw 데이터를 파이썬 리스트로 표현
- Series 클래스의 객체를 생성

````python
import pandas as pd

arr = [10, 20, 30, 40]
s = pd.Series(arr)
print(s)
````

## Series 내부 구조
■ 행 번호와 인덱스를 사용해서 데이터를 관리
- 행 번호: 자동으로 부여되는 수정이 되지 않는 번호 (print시 표시 안됨. 내부적 개념으로 알고있기)
- 인덱스: 문자 혹은 숫자로 설정 가능 (설정하지 않으면 0부터 시작하는 정수로 자동 할당 됨)

```python
data = [100, 200, 300]
s = Series(data)
print(s)
```
| 행 번호 | 인덱스 | 데이터 | 
| ------ | ---- | ---- | 
| 0 | 0 | 100 |
| 1 | 1 | 200 |
| 2 | 2 | 300 |


■ Series 생성시 index 지정하는 방법
- Series 생성시 data와 index 인자 넘겨주기

```python
data = [100, 200, 300]
index = ["월", "화", "수"]
s = Series(data, index)
print(s)
```
| 행 번호 | 인덱스 | 데이터|
| ----- | ----- | --- |
| 0 | "월" | 100 | 
| 1 | "화" | 200 |
| 2 | "수" | 300 |

````python
import pandas as pd

data = [100, 200, 300]
s = pd.Series(data)
print(s)
````

````python
import pandas as pd

data = [100, 200, 300]
index = ["월", "화", "수"]
s = pd.Series(data, index)
print(s)
````

## 📝 연습문제

### Q1. 다음 데이터를 판다스 시리즈(Series) 객체로 생성하세요. 
| index | values | 
| ----- | ------ |
| 메로나 | 500 | 
| 누가바 | 800 |
| 빠삐코 | 200 |

````python
import pandas as pd

index = ['메로나', '누가바', '빠비코']
values = [500, 800, 200]

s = pd.Series(values, index)
print(s)
````
